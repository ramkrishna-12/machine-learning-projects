## Tokenizing and Structuring Intent Data

```python
for intent in intents['intents']:
    for pattern in intent['patterns']:

        #tokenize each word
        w = nltk.word_tokenize(pattern)
        words.extend(w)
        #add documents in the corpus
        documents.append((w, intent['tag']))

        # add to our classes list
        if intent['tag'] not in classes:
            classes.append(intent['tag'])
```

This block of code processes the intent data from the JSON file by tokenizing the patterns and organizing the data into relevant lists for further use in training the chatbot model.


* `for intent in intents['intents']:`, this line loops through each intent in the intents dictionary loaded from the JSON file. Each intent contains a tag, patterns, and responses.



* `for pattern in intent['patterns']:`,this inner loop iterates through each pattern associated with the current intent. Patterns are example sentences or phrases that the user might say.


* `w = nltk.word_tokenize(pattern)`, this line tokenizes each pattern into individual words using NLTK’s word_tokenize function. Tokenization is the process of splitting text into smaller units (tokens), such as words.


* `words.extend(w)`,the tokenized words are added to the words list. This list accumulates all the words from all patterns, which will later be used to create a vocabulary.


* `documents.append((w, intent['tag']))` here each tokenized pattern, along with its corresponding intent tag, is added as a tuple to the documents list. This list represents the corpus of training data, where each document is a pair of tokenized words and their associated intent.

* `if intent['tag'] not in classes:`, this condition checks if the intent tag is already in the classes list.

* `classes.append(intent['tag'])` it the tag is not in the list, it is added. The` classes` list keeps track of all unique intent tags.

In short this code block processes the intents from the JSON file by tokenizing the patterns, extending the vocabulary list (words), creating the corpus of documents (documents), and collecting unique intent tags (classes). These structures are essential for training the chatbot model, as they provide the necessary data for understanding and categorizing user inputs.

## Creating Training Data for the Chatbot Model

```python
# create our training data
training = []

# create an empty array for our output
output_empty = [0] * len(classes)

# training set, bag of words for each sentence
for doc in documents:

    # initialize our bag of words
    bag = []
    
    # list of tokenized words for the pattern
    pattern_words = doc[0]

    # lemmatize each word - create base word, in attempt to represent related words
    pattern_words = [lemmatizer.lemmatize(word.lower()) for word in pattern_words]

    # create our bag of words array with 1, if word match found in current pattern
    for w in words:
        bag.append(1) if w in pattern_words else bag.append(0)

# output is a '0' for each tag and '1' for current tag (for each pattern)
    output_row = list(output_empty)
    output_row[classes.index(doc[1])] = 1
    training.append([bag, output_row])

# shuffle our features and turn into np.array
import numpy as np
import random

# Assuming 'words', 'classes', 'documents', and 'lemmatizer' are defined elsewhere
training = []
output_empty = [0] * len(classes)

for doc in documents:
    bag = []
    pattern_words = doc[0]
    pattern_words = [lemmatizer.lemmatize(word.lower()) for word in pattern_words]
    
    # Initialize output row inside the loop
    output_row = list(output_empty)

    for w in words:
        # Check if word is in the pattern_words
        bag.append(1) if w in pattern_words else bag.append(0)
        
    # Set the corresponding index of the output_row to 1
    output_row[classes.index(doc[1])] = 1

    # Append the bag and output_row as a tuple to training
    training.append((bag, output_row))

# Shuffle training data before converting to numpy array
random.shuffle(training)

# Print the shapes of each element in the training list
for item in training:
    print("Shape of bag:", np.array(item[0]).shape)
    print("Shape of output_row:", np.array(item[1]).shape)

# Separate features (X) and labels (Y) from the training data
train_x = np.array([bag for bag, _ in training])
train_y = np.array([output_row for _, output_row in training])

print("Training data created")



```

This code constructs the training data required to train a machine learning model for the chatbot in the following steps:

## Defining and Training the Chatbot Model


```python
# Create model - 3 layers. First layer 128 neurons, second layer 64 neurons and 3rd output layer contains number of neurons
# equal to number of intents to predict output intent with softmax
model = Sequential()
model.add(Dense(128, input_shape=(len(train_x[0]),), activation='relu'))
model.add(Dropout(0.5))
model.add(Dense(64, activation='relu'))
model.add(Dropout(0.5))
model.add(Dense(len(train_y[0]), activation='softmax'))

# Compile model. Stochastic gradient descent with Nesterov accelerated gradient gives good results for this model
#from keras.optimizers import SGD
#sgd = SGD(learning_rate=0.01, decay=1e-6, momentum=0.9, nesterov=True)

from keras.optimizers import SGD
from tensorflow.keras.optimizers.schedules import ExponentialDecay

lr_schedule = ExponentialDecay(
    initial_learning_rate=0.01,
    decay_steps=100000,
    decay_rate=0.96,
    staircase=True)

sgd = SGD(learning_rate=lr_schedule, momentum=0.9, nesterov=True)
model.compile(loss='categorical_crossentropy', optimizer=sgd, metrics=['accuracy'])

#fitting and saving the model 

hist = model.fit(np.array(train_x), np.array(train_y), epochs=200, batch_size=5, verbose=1)
model.save('chatbot_model.h5', hist)

print("model created")
```

**Compile the Model:**

* `from keras.optimizers import SGD`: Imports the Stochastic Gradient Descent optimizer.

## Processing and Predicting User Input

```python
def clean_up_sentence(sentence):
    # tokenize the pattern - split words into array
    sentence_words = nltk.word_tokenize(sentence)

    # stem each word - create short form for word
    sentence_words = [lemmatizer.lemmatize(word.lower()) for word in sentence_words]
    return sentence_words

# return bag of words array: 0 or 1 for each word in the bag that exists in the sentence
def bow(sentence, words, show_details=True):

    # tokenize the pattern
    sentence_words = clean_up_sentence(sentence)

    # bag of words - matrix of N words, vocabulary matrix
    bag = [0]*len(words) 
    for s in sentence_words:
        for i,w in enumerate(words):
            if w == s: 

                # assign 1 if current word is in the vocabulary position
                bag[i] = 1
                if show_details:
                    print ("found in bag: %s" % w)
                    
    return(np.array(bag))

def predict_class(sentence, model):
    if "bus time" in sentence.lower():
        route = None
        for word in sentence.split():
            if word.isdigit():
                route = int(word)
                break
        if route is not None:
            return [{"intent": "bus_time", "route": route}]
        else:
            return [{"intent": "noanswer"}]
    elif "stops" in sentence.lower():
        route = None
        for word in sentence.split():
            if word.isdigit():
                route = int(word)
                break
        if route is not None:
            return [{"intent": "bus_stops", "route": route}]
        else:
            return [{"intent": "noanswer"}]
    else:
        # Your existing prediction code
        p = bow(sentence, words, show_details=False)
        res = model.predict(np.array([p]))[0]
        ERROR_THRESHOLD = 0.25
        results = [[i,r] for i,r in enumerate(res) if r>ERROR_THRESHOLD]
        results.sort(key=lambda x: x[1], reverse=True)
        return_list = []
        for r in results:
            return_list.append({"intent": classes[r[0]], "probability": str(r[1])})
        return return_list



```

In this block of code, three functions are created to process user input and predict the intent using the trained chatbot mode:

**1. clean_up_sentence**:This function tokenizes a given sentence into individual words and then lemmatizes each word to its base form. This helps in normalizing the words by converting them to their root forms.

**2. bow (Bag of Words)**: This function creates a bag of words (BoW) representation of a given sentence. It initializes a list with zeros and sets the position to 1 if the word exists in the sentence. This representation helps in mapping the sentence to a fixed-length vector based on the known vocabulary.
 
 **3. predict_class**:This function predicts the intent of a given sentence using the trained model. It first checks for specific keywords like "bus time" and "stops" to directly identify the route number. For other queries, it uses the bag of words representation to predict the intent using the trained model and returns a list of intents with their respective probabilities


Here the 3 functions are discussed in details:

**1. clean_up_sentence:**

   * **Tokenization**: This function takes a sentence as input and uses the nltk.word_tokenize function to split the sentence into individual words or tokens, storing them in the list sentence_words.

  * **Lemmatization**: The function then lemmatizes each token using the lemmatizer.lemmatize method, which converts each word to its base form (e.g., "running" becomes "run"). This normalized list of tokens is returned by the function.

**2. bow (Bag of Words):**

* **Tokenization and Lemmatization:**

   * The function starts by calling `clean_up_sentence` on the input sentence. This process tokenizes the sentence into individual words and lemmatizes each word to its base form, resulting in a list `sentence_words`.

* **Creating the Bag of Words:**

    * A list `bag of zeros` is initialized, with a length equal to the number of words in the vocabulary (words). This list represents the bag of words, where each position corresponds to a word in the vocabulary.
    * The function then iterates over each word s in the tokenized and lemmatized sentence (sentence_words). For each word s, it checks if this word is present in the vocabulary (words).
    * If a match is found (i.e., w == s), the corresponding position in the bag list is set to 1, indicating the presence of that word in the sentence. If show_details is True, the function prints out each word found in the bag.
    * Finally, the bag list, now a binary representation of the presence of words from the vocabulary in the input sentence, is converted to a NumPy array and returned.
    
**3. predict_class:**

* **Identify Specific Requests:**

  * The function checks if the input `sentence` contains the phrase "bus time". If found, it attempts to extract a route number (a digit) from the sentence. If a route number is identified, it returns a dictionary indicating the intent "bus_time" along with the route number. If no route number is found, it returns an intent of "noanswer".
  * Similarly, it checks if the sentence contains the word "stops". If found, it follows the same process to extract a route number and returns a dictionary with the intent "bus_stops" and the route number. If no route number is found, it returns an intent of "noanswer".

* **General Intent Prediction:**

  * If the input sentence doesn't contain the specific phrases "bus time" or "stops", the function uses the trained model to predict the general intent of the sentence.
  * The function converts the sentence into a bag-of-words representation using the bow function.
  * It then feeds this representation into the model to get a list of probabilities for each possible intent.
  * It filters out intents with probabilities below a certain threshold `(ERROR_THRESHOLD)`, sorts the remaining intents by probability in descending order, and constructs a list of dictionaries with the predicted intents and their associated probabilities.

## Generating Chatbot Responses and Extracting Bus Route Information

```python
def getResponse(ints, intents_json):
    tag = ints[0]['intent']
    list_of_intents = intents_json['intents']
    for i in list_of_intents:
        if i['tag'] == tag:
            result = random.choice(i['responses'])
            break
    return result

def chatbot_response(text):
    ints = predict_class(text, model)
    for intent in ints:
        if intent['intent'] == 'bus_stops':
            route = intent['route']
            stops = get_bus_stops(route, intents)
            if stops:
                return stops
            else:
                return "Sorry, I couldn't find stops for Route {}.".format(route)
    res = getResponse(ints, intents)
    return res


def get_bus_stops(route, intents_json):
    for intent in intents_json['intents']:
        for response in intent['responses']:
            if response.startswith(f"The stops for Route {route} are"):
                stops_part = response.split(":")[1].strip()
                stops_list = [stop.strip() for stop in stops_part.split(",")]
                stops_str = ", ".join(stops_list)
                return stops_str
    return None


def extract_stops(responses):
    for response in responses:
        if response.startswith("The stops for"):
            stops_part = response.split(":")[1].strip()
            stops_list = [stop.strip() for stop in stops_part.split(",")]
            stops_str = ", ".join(stops_list)
            return stops_str
    return None

```

This code block includes functions to handle chatbot responses, particularly for providing bus stop information for specific routes. It integrates intent prediction, response generation, and extraction of bus stop details from a predefined set of responses.

Here 4 functions are created and they are discussed below: 

**1. Function getResponse:**

* This function takes the predicted intents (`ints`) and the JSON data containing the intents `(intents_json)`.
* It identifies the tag of the highest probability intent (`ints[0]['intent']`).
* It then iterates through the list of intents in `intents_json` to find the matching intent.
* Once found, it randomly selects a response from the list of responses associated with that intent and returns it.

**2.Function chatbot_response:**

* This function takes user input text (`text`), predicts the intent using the `predict_class` function, and retrieves the appropriate response.
* If the predicted intent is 'bus_stops', it extracts the route number and calls `get_bus_stops` to get the stops for that route.
* If stops are found, it returns them. If not, it returns an error message.
* For other intents, it uses `getResponse` to fetch a response.

## Creating a Chatbot GUI with Tkinter

```python
#Creating GUI with tkinter

import tkinter
from tkinter import *

def send():
    msg = EntryBox.get("1.0",'end-1c').strip()
    EntryBox.delete("0.0",END)

    if msg != '':
        ChatLog.config(state=NORMAL)
        ChatLog.insert(END, "You: " + msg + '\n\n')
        ChatLog.config(foreground="#442265", font=("Verdana", 12 ))
        res = chatbot_response(msg)
        ChatLog.insert(END, "Bot: " + res + '\n\n')
        ChatLog.config(state=DISABLED)
        ChatLog.yview(END)

base = Tk()
base.title("Hello")
base.geometry("400x500")
base.resizable(width=FALSE, height=FALSE)

#Create Chat window
ChatLog = Text(base, bd=0, bg="white", height="8", width="50", font="Arial",)
ChatLog.config(state=DISABLED)

#Bind scrollbar to Chat window
scrollbar = Scrollbar(base, command=ChatLog.yview, cursor="heart")
ChatLog['yscrollcommand'] = scrollbar.set

#Create Button to send message
SendButton = Button(base, font=("Verdana",12,'bold'), text="Send", width="12", height=5,
                    bd=0, bg="#32de97", activebackground="#3c9d9b",fg='#ffffff',
                    command= send )

#Create the box to enter message
EntryBox = Text(base, bd=0, bg="white",width="29", height="5", font="Arial")

#EntryBox.bind("<Return>", send)
#Place all components on the screen
scrollbar.place(x=376,y=6, height=386)
ChatLog.place(x=6,y=6, height=386, width=370)
EntryBox.place(x=128, y=401, height=90, width=265)
SendButton.place(x=6, y=401, height=90)
base.mainloop()
```

**Import Libraries:**

* The code begins by importing the `tkinter` library and all its components (*), which are used to build the GUI.

**Function `send`**:

* This function handles the sending of messages.
* It retrieves the text entered by the user in the `EntryBox`, clears the EntryBox, and displays the user's message in the `ChatLog`.
* It then gets a response from the `chatbot_response` function and displays it in the `ChatLog`.

**Initialize the Main Window**:

* The main window (`base`) is created, with a title "Hello", a fixed size of 400x500 pixels, and it is not resizable.

**Create Chat Window:**

A `Text` widget named `ChatLog` is created for displaying the chat history. It is initially disabled to prevent user input directly in the chat log.

**Bind Scrollbar:**

* A `Scrollbar` widget is created and bound to the `ChatLog` to allow scrolling through the chat history.

**Create Send Button:**

* A `Button` widget named `SendButton` is created for sending messages. When clicked, it triggers the `send` function.

**Create Message Entry Box:**

* A `Text` widget named `EntryBox` is created for users to type their messages.

**Place GUI Components:**

* The `scrollbar, ChatLog, EntryBox`, and `SendButton `are placed on the screen using the `place` method with specified coordinates.

**Main Loop:**

* `base.mainloop()` starts the main event loop, waiting for user interactions.

in short this code sets up a simple GUI for a chatbot using tkinter, where users can input messages, send them using a button, and see both their messages and the chatbot's responses in a chat window.
